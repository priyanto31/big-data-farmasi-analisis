```python
import pandas as pd
import sqlite3

# 1. Load dataset ke dalam Pandas DataFrame
print("Memuat dataset...")
df = pd.read_csv('Dataset_Farmasi_Priyanto_15240017.csv')

# Menampilkan 5 baris pertama untuk memastikan data terbaca dengan benar
print("\n--- 5 Baris Pertama Data ---")
display(df.head())

# Profiling informasi dasar (tipe data & mengecek apakah ada data yang kosong/missing)
print("\n--- Informasi Struktur Data ---")
df.info()

# Statistik deskriptif untuk kolom angka (kuantitas, harga, total nilai)
print("\n--- Statistik Deskriptif ---")
display(df.describe())

# 2. Uji Coba Query SQL menggunakan SQLite in-memory
print("\nMenyiapkan database SQL sementara...")
conn = sqlite3.connect(':memory:')
df.to_sql('farmasi_transaksi', conn, index=False, if_exists='replace')

# Query untuk melihat Gudang dan Kategori obat apa yang menghasilkan pendapatan tertinggi
query = """
    SELECT warehouse, product_category, COUNT(*) as total_transaksi, SUM(total_value) as total_pendapatan
    FROM farmasi_transaksi
    GROUP BY warehouse, product_category
    ORDER BY total_pendapatan DESC
    LIMIT 5;
"""
print("\n--- Hasil Query SQL: Top 5 Pendapatan per Gudang & Kategori ---")
df_sql = pd.read_sql(query, conn)
display(df_sql)
```
